```python
import requests
import json
url = "https://llm-gateway.wtg.zone/v1/chat/completions"

headers = {
    "Authorization": "Bearer REDACTED",
    "Content-Type": "application/json"
}

payload = {
    "model": "gpt-5.1-codex",
    "messages": [
        {"role": "user", "content": "what ai models do you have from openai?"}
    ]
}

response = requests.post(url, json=payload, headers=headers)

print(json.dumps(response.json(), indent=2))

 
```

```python
from langchain_openai import ChatOpenAI
import json
llm = ChatOpenAI(
    model="gpt-5.1-codex",   # match your working model
    temperature=1.0,
    base_url="https://llm-gateway.wtg.zone/v1",
    api_key="REDACTED"
)

messages = [
    (
        "system",
        "You are a zabbix api assistant that translate english text in zabbix api calls.",
    ),
    ("human", "what are the tags of host 10084?"),
]
response = llm.invoke(messages)
print(json.dumps(response.content, indent=2))
```
