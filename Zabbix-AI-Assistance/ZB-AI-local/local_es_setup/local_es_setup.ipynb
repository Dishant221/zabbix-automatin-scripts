Index mapping for rag design

| Field              | Purpose                    |
| ------------------ | -------------------------- |
| api_name           | API method                 |
| method             |  zabbix api method         |
| chunk_type         | usecase/parameter/overview |
| title              | human title                |
| description        | semantic meaning           |
| content            | full semantic text         |
| parameter_name     | name of the parameter      |
| parameter_type     | type of parameter          |
| usecase            | what is the usecase        |
| keywords           | retrieval helpers          |
| example_request    | API example                |
| example_response   | optional                   |
| related_parameters | related params             |
| common_usecases    | usecases                   |
| metadata           | flexible extension         |
| embedding_text     | embedding input            |
| embedding          | vector                     |


```python
from elasticsearch import Elasticsearch

es = Elasticsearch(
    "http://localhost:9200",
    request_timeout=30
)

index_name = "zabbix_api"

mapping_format = {
    "mappings": {
        "properties": {

            "api_name": {
                "type": "keyword"
            },

            "method": {
                "type": "keyword"
            },

            "chunk_type": {
                "type": "keyword"
            },

            "title": {
                "type": "text"
            },

            "description": {
                "type": "text"
            },

            "content": {
                "type": "text"
            },

            "parameter_name": {
                "type": "keyword"
            },

            "parameter_type": {
                "type": "keyword"
            },

            "usecase": {
                "type": "keyword"
            },

            "purpose": {
                "type": "text"
            },

            "how_it_works": {
                "type": "text"
            },

            "usage_pattern": {
                "type": "text"
            },

            "required_parameters": {
                "type": "object",
                "enabled": True
            },

            "optional_parameters": {
                "type": "keyword"
            },

            "important_parameters": {
                "type": "object",
                "enabled": True
            },

            "response_fields": {
                "type": "keyword"
            },

            "supported_values": {
                "type": "keyword"
            },

            "common_usecases": {
                "type": "keyword"
            },

            "keywords": {
                "type": "keyword"
            },

            "example_request": {
                "type": "object",
                "enabled": True
            },

            "example_response": {
                "type": "object",
                "enabled": True
            },

            "metadata": {
                "type": "object",
                "enabled": True
            },

            "embedding_text": {
                "type": "text"
            },

            "embedding": {
                "type": "dense_vector",
                "dims": 384,
                "index": True,
                "similarity": "cosine"
            }
        }
    }
}



if not es.indices.exists(index=index_name):
    es.indices.create(index=index_name, body=mapping_format)
    print(f"Index '{index_name}' created successfully.")
else:
    print(f"Index '{index_name}' already exists.")
```

IMPORTANT DESIGN RULE

You should ALSO create parameter chunks separately.

Because user may ask:

"What does selectInterfaces do?"

NOT a usecase.

***Parameter Chunks***

question for parameters 

For API RAG, parameter chunks are EXTREMELY important because users often ask:

“How to use selectInterfaces?”
“What does limitSelects do?”
“How to filter hosts?”
“How does searchInventory work?”

So every important parameter should become its own semantic chunk


*Parameter Chunk Structure

Every parameter chunk should contain:*

```python
{
  "api_name": "host.get",
  "chunk_type": "parameter",
  "parameter_name": "selectInterfaces",
  "purpose": "",
  "parameter_type": "",
  "description": "",
  "how_it_works": "",
  "supported_values": [],
  "usage_pattern": "",
  "api_example": {},
  "response_example": {},
  "common_usecases": [],
  "keywords": []
}
```
