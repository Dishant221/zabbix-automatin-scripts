```python
import requests
import json
url = "https://llm-gateway.wtg.zone/v1/chat/completions"

headers = {
    "Authorization": "Bearer REDACTED",
    "Content-Type": "application/json"
}

payload = {
    "model": "gpt-5.1-codex",
    "messages": [
        {"role": "user", "content": "if i give you zabbix dumnetation and ask you u to generate zabbix api call ,, can u do that ?"}
    ]
}

response = requests.post(url, json=payload, headers=headers)

print(json.dumps(response.json(), indent=2))
```

```python
from langchain.chat_models import init_chat_model
import os
import json

os.environ["OPENAI_API_KEY"] = "REDACTED"

model = init_chat_model(
    "gpt-5.1-codex",
    model_provider="openai",
    base_url="https://llm-gateway.wtg.zone/v1",
    temperature=0.1,

)
answer = model.invoke("What is PyCharm?")
print(json.dumps(answer.dict(), indent=2))
```
