```python
def api_get(token, endpoint, page_size=100, page_number=1):
    url = f"{api_url}{endpoint}"
    headers = {'Authorization': f'Bearer {token}'}
    params = {'page_size': page_size, 'page_number': page_number}
    resp = requests.get(url, headers=headers, params=params)
    resp.raise_for_status()
    return resp.json()
```

```python
Total_Licences_Records = 1036
def get_license_metrics(token, page_size=100):
    all_license_items = []
    total_pages = (Total_Licences_Records + page_size - 1) // page_size
    for page_number in range(1, total_pages + 1):
        data = api_get(token, "/api/sam/v1/licenses", page_size, page_number)
        items = data.get('items', [])
        if not items:
            break
        all_license_items.extend(items)
        print(f"Page {page_number}/{total_pages} fetched - Total so far: {len(all_license_items)}")
    print(f"\nTotal License Metrics fetched: {len(all_license_items)}")
    with open('licenses.json', 'w') as f:
        json.dump(all_license_items, f, indent=2)
    print(f"Saved to licenses.json")
    return all_license_items

get_license_metrics(token)
```
